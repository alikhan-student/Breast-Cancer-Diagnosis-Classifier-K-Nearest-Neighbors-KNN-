```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv('/breast-cancer.csv')
print(df.head)
print(df.info)
print(df.columns)
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.drop('diagnosis',axis=1),df['diagnosis'],test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train,y_train)
y_pred=knn.predict(x_test)
accuracy=accuracy_score(y_test,y_pred)
print('Accuracy:',accuracy)
```

```python
accuracy=[]
for i in range(1,21):
    knn=KNeighborsClassifier(n_neighbors=i)
    knn.fit(x_train,y_train)
    y_pred=knn.predict(x_test)
    accuracy.append(accuracy_score(y_test,y_pred))
```

```python
figure=plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
plt.pie(accuracy,labels=range(1,21),autopct='%1.1f%%')
plt.title('Pie Chart')
plt.subplot(2,2,2)
plt.bar(range(1,21),accuracy)
plt.xlabel('K')
plt.ylabel('Accuracy')
plt.title('Bar Chart')
plt.subplot(2,2,3)
plt.scatter(range(1,21),accuracy)
plt.xlabel('K')
plt.ylabel('Accuracy')
plt.title('Scatter Plot')
plt.subplot(2,2,4)
plt.plot(range(1,21),accuracy)
plt.xlabel('K')
plt.ylabel('Accuracy')
plt.title('Line Chart')
plt.show()
```

### Visualizing KNN Decision Boundary

To visualize the KNN model's decision boundary, we need to reduce the data to two dimensions. For this demonstration, we will use `radius_mean` and `texture_mean` as the features.

```python
# Select two features for visualization
X_viz = df[['radius_mean', 'texture_mean']]
y_viz = df['diagnosis']

# Scale the features (important for KNN)
scaler_viz = StandardScaler()
X_viz_scaled = scaler_viz.fit_transform(X_viz)

# Convert 'M' and 'B' to numerical values for plotting decision boundary
y_viz_numeric = y_viz.map({'M': 1, 'B': 0})

# Train a KNN classifier on these two features (using an optimal k, e.g., k=9 from previous analysis)
knn_viz = KNeighborsClassifier(n_neighbors=9)
knn_viz.fit(X_viz_scaled, y_viz_numeric)

# Create a meshgrid for plotting the decision boundary
x_min, x_max = X_viz_scaled[:, 0].min() - 1, X_viz_scaled[:, 0].max() + 1
y_min, y_max = X_viz_scaled[:, 1].min() - 1, X_viz_scaled[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                     np.arange(y_min, y_max, 0.02))

# Predict on the meshgrid to get decision regions
Z = knn_viz.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot the decision boundary and the data points
plt.figure(figsize=(10, 8))
plt.contourf(xx, yy, Z, alpha=0.8, cmap=plt.cm.RdYlBu)
plt.scatter(X_viz_scaled[:, 0], X_viz_scaled[:, 1], c=y_viz_numeric, cmap=plt.cm.RdYlBu, edgecolors='k', s=50)

plt.xlabel('Scaled Radius Mean')
plt.ylabel('Scaled Texture Mean')
plt.title('KNN Decision Boundary (k=9) on Radius Mean vs Texture Mean')
plt.show()
```
